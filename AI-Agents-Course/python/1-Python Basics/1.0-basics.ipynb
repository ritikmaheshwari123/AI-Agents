```python
## Basic Syntax rules for Python
## Case sensitivity - Python is case sensitive

name = "Ritik"
Name = "Maheshwari"

print(name)  
print(Name)  
```

```python
## Indentation
## Python uses identation to define blocks of code. Consistent use of spaces (commonly 4) or a tab is required.

age = 27
if age >= 18:
    print(age)
```

```python
## This is a single line comment
print("Hello, World!")
```

```python
## Line continuation
## Use a backslash (\) to continue the statement to the next line.

total = 1 + 2 + 3 + \
4 + 5
print(total)
```

```python
## Multiple statements on a single line

x=8; y=10; z=x+y
print(z)
```

```python
## Understand semantics in python
## Variable assignments

age=27
name="Ritik"
```
